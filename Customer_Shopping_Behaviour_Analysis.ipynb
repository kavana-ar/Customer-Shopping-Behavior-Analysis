```python
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Purchase Amount (USD)'] = df.groupby('Item Purchased')['Purchase Amount (USD)'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Previous Purchases'] = df.groupby('Item Purchased')['Previous Purchases'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Size'] = df.groupby('Category')['Size'].transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else df['Size'].mode().iloc[0]))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create column purchase _frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase _frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase _frequency_days','frequency_of_purchases']].head(10)
```
